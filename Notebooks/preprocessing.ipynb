```python
import pandas as pd
import numpy as np

data = pd.read_csv("../Dataset/Car Price.csv")

print("Dataset loaded successfully!")
print("Original shape:", data.shape)

data.head()
```

```python
print("Duplicate rows before removal:", data.duplicated().sum())

data = data.drop_duplicates().reset_index(drop=True)

print("Duplicate rows after removal:", data.duplicated().sum())
print("New dataset shape:", data.shape)
```

```python
print("Data Types:")
print(data.dtypes)
```

```python
from datetime import datetime

current_year = datetime.now().year

data["Car_Age"] = current_year - data["Year"]

data["KM_per_Year"] = (
    data["KM_Driven"] /
    data["Car_Age"].replace(0, 1)
)

print("Current Year:", current_year)
print("Feature engineering completed!")

data[
    ["Year", "KM_Driven", "Car_Age", "KM_per_Year"]
].head()
```

```python
features = [
    "Brand",
    "Model",
    "Year",
    "KM_Driven",
    "Fuel",
    "Seller_Type",
    "Transmission",
    "Owner",
    "Car_Age",
    "KM_per_Year"
]

X = data[features]

y = data["Selling_Price"]

print("Features selected successfully!")

print("\nInput Features:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)

print("\nFeature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
categorical_features = [
    "Brand",
    "Model",
    "Fuel",
    "Seller_Type",
    "Transmission",
    "Owner"
]

numerical_features = [
    "Year",
    "KM_Driven",
    "Car_Age",
    "KM_per_Year"
]

print("Categorical Features:")
print(categorical_features)

print("\nNumerical Features:")
print(numerical_features)
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

encoded_data = encoder.fit_transform(
    X[categorical_features]
)

print("One-Hot Encoding completed!")
print("Encoded shape:", encoded_data.shape)
```

```python
# Convert numerical features to NumPy array
numerical_data = X[numerical_features].to_numpy()

# Combine categorical and numerical features
X_processed = np.hstack([
    encoded_data,
    numerical_data
])

print("Preprocessing completed!")
print("Final processed feature shape:", X_processed.shape)
print("Target shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_processed,
    y,
    test_size=0.20,
    random_state=42
)

print("Train-Test Split completed!")

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
print("FEATURE ENGINEERING SUMMARY")
print("=" * 40)

print("1. Car_Age")
print("   Formula: Current Year - Year")

print("\n2. KM_per_Year")
print("   Formula: KM_Driven / Car_Age")

print("\nEngineered Features:")
print(data[[
    "Year",
    "KM_Driven",
    "Car_Age",
    "KM_per_Year"
]].head())
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("ML models and evaluation metrics imported successfully!")
```

```python
from sklearn.model_selection import GridSearchCV

print("GridSearchCV imported successfully!")
```

```python
rf = RandomForestRegressor(
    random_state=42,
    n_jobs=-1
)

param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

print("Hyperparameter grid created!")
print("Total combinations:",
      2 * 3 * 2 * 2)
```

```python
rf = RandomForestRegressor(
    random_state=42,
    n_jobs=-1
)

param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

print("Hyperparameter grid created!")
print("Total combinations:",
      2 * 3 * 2 * 2)
```

```python
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=3,
    scoring="r2",
    n_jobs=-1,
    verbose=1
)
print("Starting hyperparameter tuning...")
grid_search.fit(X_train, y_train)
print("Hyperparameter tuning completed!")
```

```python
print("BEST HYPERPARAMETERS")
print("=" * 40)

print(grid_search.best_params_)

print("\nBest Cross-Validation R²:")
print(grid_search.best_score_)
```

```python
best_rf = grid_search.best_estimator_

tuned_pred = best_rf.predict(X_test)

tuned_mae = mean_absolute_error(y_test, tuned_pred)
tuned_mse = mean_squared_error(y_test, tuned_pred)
tuned_rmse = np.sqrt(tuned_mse)
tuned_r2 = r2_score(y_test, tuned_pred)

print("TUNED RANDOM FOREST RESULTS")
print("=" * 40)

print("MAE :", round(tuned_mae, 2))
print("MSE :", round(tuned_mse, 2))
print("RMSE:", round(tuned_rmse, 2))
print("R²  :", round(tuned_r2, 4))
```

```python
print("RANDOM FOREST - BEFORE vs AFTER TUNING")
print("=" * 50)

print("Before Tuning")
print("R² :", 0.6936)
print("MAE:", 102623.16)

print("\nAfter Tuning")
print("R² :", round(tuned_r2, 4))
print("MAE:", round(tuned_mae, 2))

print("\nFINAL DECISION")

if tuned_r2 > 0.6936:
    print("Tuned Random Forest selected.")
else:
    print("Original Random Forest selected.")
```
