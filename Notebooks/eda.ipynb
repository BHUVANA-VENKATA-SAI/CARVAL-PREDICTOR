```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
data = pd.read_csv("../Dataset/Car Price.csv")

print("Dataset loaded successfully!")
print("Shape:", data.shape)

data.head()
```

```python
plt.figure(figsize=(10, 5))

plt.hist(data["Selling_Price"], bins=30)

plt.title("Distribution of Selling Price")
plt.xlabel("Selling Price (₹)")
plt.ylabel("Number of Cars")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    data["Year"],
    data["Selling_Price"],
    alpha=0.5
)

plt.title("Selling Price vs Manufacturing Year")
plt.xlabel("Manufacturing Year")
plt.ylabel("Selling Price (₹)")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    data["KM_Driven"],
    data["Selling_Price"],
    alpha=0.5
)

plt.title("Selling Price vs Kilometers Driven")
plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price (₹)")

plt.show()
```

```python
fuel_price = data.groupby("Fuel")["Selling_Price"].mean()

print(fuel_price)

plt.figure(figsize=(8, 5))

fuel_price.plot(kind="bar")

plt.title("Average Selling Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average Selling Price (₹)")

plt.xticks(rotation=0)

plt.show()
```

```python
transmission_price = data.groupby("Transmission")["Selling_Price"].mean()

print(transmission_price)

plt.figure(figsize=(8, 5))

transmission_price.plot(kind="bar")

plt.title("Average Selling Price by Transmission")
plt.xlabel("Transmission Type")
plt.ylabel("Average Selling Price (₹)")

plt.xticks(rotation=0)

plt.show()
```
